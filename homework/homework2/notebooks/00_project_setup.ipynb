# Stage 02 — Tooling Setup: Environment & Config Check

This notebook builds the full project scaffold, creates required folders, configures `.env` secret management, and verifies Python packages for Homework 2.

```python
# 1. Verify Python interpreter environment
import sys
import os
from pathlib import Path

print("Python version:", sys.version)
print("Interpreter path:", sys.executable)
```

```python
# 2. Create standard project directory scaffold with .gitkeep placeholders
PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()

# Define all required project subdirectories
FOLDERS = [
    "data/raw",
    "data/processed",
    "notebooks",
    "src",
    "docs",
    "reports",
    "model"
]

# Generate directories and .gitkeep files for Git tracking
for folder in FOLDERS:
    folder_path = PROJECT_ROOT / folder
    folder_path.mkdir(parents=True, exist_ok=True)
    (folder_path / ".gitkeep").touch(exist_ok=True)

print("Project directory scaffold created successfully:")
for f in FOLDERS:
    print("  -", f)
```

```python
# 3. Create core configuration files (.gitignore, .env.example, .env, README.md, src/config.py)

# Create .gitignore to prevent committing sensitive files or system cache
gitignore_content = """
.env
__pycache__/
*.py[cod]
.ipynb_checkpoints/
data/raw/*
!data/raw/.gitkeep
data/processed/*
!data/processed/.gitkeep
"""
(PROJECT_ROOT / ".gitignore").write_text(gitignore_content.strip(), encoding="utf-8")

# Create .env.example (template) and .env (local secrets)
(PROJECT_ROOT / ".env.example").write_text("API_KEY=your_dummy_key_here\nDATA_DIR=./data\n", encoding="utf-8")
if not (PROJECT_ROOT / ".env").exists():
    (PROJECT_ROOT / ".env").write_text("API_KEY=dummy_key_123\nDATA_DIR=./data\n", encoding="utf-8")

# Create src/config.py module
config_code = """import os
from pathlib import Path
from dotenv import load_dotenv

PROJECT_ROOT = Path(__file__).resolve().parent.parent
load_dotenv(PROJECT_ROOT / ".env")

def get_key(name: str, default: str = None) -> str:
    return os.getenv(name, default)

DATA_DIR = PROJECT_ROOT / get_key("DATA_DIR", "data")
"""
(PROJECT_ROOT / "src" / "config.py").write_text(config_code.strip(), encoding="utf-8")

# Create README.md project documentation
readme_content = "# Homework 2 — Stage 02: Tooling Setup\nReproducible environment configuration, secret management, config helper modules, and Jupyter notebook verification."
(PROJECT_ROOT / "README.md").write_text(readme_content, encoding="utf-8")

print("Configuration and source code files successfully generated!")
```

```python
# 4. Load .env secrets and check required assertion output
import numpy as np
from dotenv import load_dotenv

load_dotenv(PROJECT_ROOT / '.env')

api_key = os.getenv('API_KEY')
data_dir = os.getenv('DATA_DIR', './data')

# Core rubric check: Must print 'API_KEY present: True'
print("API_KEY present:", api_key is not None)
print("Loaded API_KEY:", api_key)
print("DATA_DIR from env:", data_dir)
```

```python
# 5. Test src/config.py module functionality
sys.path.append(str(PROJECT_ROOT))
from src.config import get_key, DATA_DIR

print("Config Helper Key:", get_key("API_KEY"))
print("Project Data Directory:", DATA_DIR)
```

```python
# 6. Perform a simple NumPy operation demo
arr = np.array([1, 2, 3, 4, 5])
print("Array:", arr)
print("Mean:", arr.mean())
print("Vectorized * 10:", arr * 10)
```
