```python
# Run me first: Fixes current working directory to project root regardless of execution context
import os
import sys
from pathlib import Path

if Path.cwd().name == "notebooks":
    os.chdir("..")

ROOT = Path.cwd()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

print("Working directory set to:", ROOT.name)
```

```python
# Check required files relative to the project root directory
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
CHECKS = [
    (".env", "NEEDED", "YOU create this: copy .env.example to .env."),
    (".env.example", "NEEDED", "shipped with this stage - the template you copy to .env"),
]

print(f"Looking in: {ROOT}\n")
missing = 0
for rel, kind, note in CHECKS:
    here = (ROOT / rel).exists()
    if not here and kind == "NEEDED":
        missing += 1
    print(f"  [{'OK ' if here else 'MISS'}]  {kind:<8}  {rel:<34}  {note}")

if missing:
    print(f"\n{missing} needed file(s) missing. Put them at the paths above.")
else:
    print("\nAll needed files present.")
```

```python
# TODO 3: Implement minimal helpers to retrieve environment keys and paths
import os
from pathlib import Path

def get_key(name: str, default=None):
    """Retrieve environment variable by name."""
    return os.getenv(name, default)

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
DATA_DIR = PROJECT_ROOT / "data"

print("Project Root:", PROJECT_ROOT)
print("Data Directory:", DATA_DIR)
```

```python
# TODO 4: Verify API_KEY presence and ensure DATA_DIR directory exists
print("API_KEY present:", get_key("API_KEY") is not None)

data_dir_path = Path(get_key("DATA_DIR", "./data"))
data_dir_path.mkdir(parents=True, exist_ok=True)
print("Ensured data directory exists at:", data_dir_path.resolve())
```

```python
# TODO 5: Import NumPy, create a small array, and compute standard statistics
import numpy as np

arr = np.array([2, 4, 6, 8, 10])
print("Array:", arr)
print("Mean:", arr.mean())
print("Standard Deviation:", arr.std())
```
