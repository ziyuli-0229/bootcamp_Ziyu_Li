# Stage 01: Problem Framing & Scoping — ETF Strategy Viability 

## Stakeholder Context Artifact
> **Core Question:** *“Will this ETF factor strategy still work next year?*                                              
> **Decision Owner:** Portfolio Manager (PM)  
> **Primary Users:** PM & Quantitative Analysts  
> **Objective:** Determine whether to maintain capital allocation, adjust parameter weights, or de-allocate from the current ETF strategy for the upcoming fiscal year.

## Project Scoping Paragraph & Framing

### Problem Statement
Quantitative trading strategies frequently experience alpha decay and performance degradation due to evolving market regimes, macroeconomic shifts, and factor crowding. Our proprietary multi-factor ETF momentum strategy has generated risk-adjusted returns over the past three years; however, changing interest rate expectations and liquidity dynamics raise serious concerns regarding its persistence for the upcoming 12 months. Unintentionally continuing an underperforming strategy risks capital drawdowns and opportunity costs for the fund. This project establishes an empirical evaluation framework to stress-test the strategy, forecast out-of-sample risk/return profiles, and inform capital allocation decisions prior to the next annual rebalancing window.

### Stakeholder & User
- **Decision Owner (Stakeholder):** Portfolio Manager (PM), who holds ultimate fiduciary responsibility for fund performance and allocation decisions.
- **End Users / Operators:** Quantitative Analysts and Risk Officers who execute model diagnostics, run monthly backtest updates, and monitor risk threshold triggers.
- **Timing & Workflow Context:** Rebalance evaluation takes place quarterly, requiring diagnostic outputs 5 trading days prior to execution.

### Useful Answer & Decision
- **Framing Type:** Predictive & Descriptive.
- **Evaluation Metrics:** Out-of-sample Sharpe Ratio (> 0.8), Maximum Drawdown (< 15%), Information Ratio relative to benchmark (> 0.5), and Tracking Error.
- **Target Artifact:** Automated ETF Performance & Risk Assessment Notebook + Executive Stakeholder Memo (`docs/stakeholder_memo.md`).
- **Decision Trigger:** If forecasted out-of-sample Sharpe Ratio drops below 0.5 or simulated Max Drawdown exceeds 15% under regime-stress scenarios, trigger strategy de-allocation or risk-budget reduction.

### Assumptions & Constraints
- **Assumptions:** Asset liquidity remains sufficient for execution without excessive slippage; factor risk premiums persist across rolling 12-month windows; historical price/volume data is free of survivorship bias.
- **Constraints:** Transaction costs set at 10–15 bps per trade; maximum fund capacity capped at $50M AUM; rebalance turnover limit < 20% per cycle.

### Known Unknowns / Risks
- Macroeconomic regime shifts (e.g., unexpected central bank interest rate pivots).
- Factor crowding leading to sudden liquidity crunches and execution drag.
- Model overfitting and stationarity breakdown in rolling factor correlations.

### Lifecycle Mapping
- **Goal 1:** Scope ETF strategy viability problem $\rightarrow$ **Stage 01:** Problem Framing & Scoping $\rightarrow$ **Deliverable:** `README.md` & `docs/stakeholder_memo.md`
- **Goal 2:** Ingest and clean ETF price/volume series $\rightarrow$ **Stage 02:** Data Acquisition $\rightarrow$ **Deliverable:** `src/data_ingestion.py` & cleaned data in `data/processed/`
- **Goal 3:** Perform factor EDA & backtest evaluation $\rightarrow$ **Stage 03:** EDA & Feature Engineering $\rightarrow$ **Deliverable:** `notebooks/01_eda_backtest.ipynb`
- **Goal 4:** Forecast out-of-sample regime performance $\rightarrow$ **Stage 04:** Modeling & Validation $\rightarrow$ **Deliverable:** Risk/return prediction model
- **Goal 5:** Deploy monitoring dashboard & decision memo $\rightarrow$ **Stage 05:** Deployment $\rightarrow$ **Deliverable:** Interactive report & dashboard
